## Filtering Data

```python
# Convert Date to datetime for filtering
df['Date'] = pd.to_datetime(df['Date'])
df['Month'] = df['Date'].dt.month
df['Month_Name'] = df['Date'].dt.month_name()
df['Day_Name'] = df['Date'].dt.day_name()
```

```python
# --- Exercise 2.1: Simple Filtering ---
print("\n--- Exercise 2.1: Simple Filtering (Single Condition) ---")

# Filter Electronics products
electronics = df[df['Product Category'] == 'Electronics']
print(f"\n🔍 Electronics Products:")
print(f"   Total transactions: {len(electronics)}")
print(f"   Total revenue: Rp {electronics['Total Revenue'].sum():,.2f}")
print(electronics[['Transaction ID', 'Product Name', 'Total Revenue']].head())

# Filter high-value transactions (> 5,000,000)
high_value = df[df['Total Revenue'] > 5000000]
print(f"\n🔍 High Value Transactions (> USD 5,000,000):")
print(f"   Count: {len(high_value)}")
print(high_value[['Transaction ID', 'Product Category', 'Total Revenue']].head())

# Filter Jakarta region
jakarta_sales = df[df['Region'] == 'Jakarta']
print(f"\n🔍 Jakarta Sales:")
print(f"   Transactions: {len(jakarta_sales)}")
print(f"   Total revenue: USD {jakarta_sales['Total Revenue'].sum():,.2f}")
```

```python
# --- Exercise 2.2: Multiple Filters with AND (&) ---
print("\n--- Exercise 2.2: Multiple Filters with AND (&) ---")

# Electronics in Jakarta
electronics_jakarta = df[(df['Product Category'] == 'Electronics') &
                         (df['Region'] == 'Jakarta')]
print(f"\n🔍 Electronics in Jakarta:")
print(f"   Count: {len(electronics_jakarta)}")
print(electronics_jakarta[['Product Category', 'Region', 'Total Revenue']].head())

# High value Credit Card transactions
high_credit = df[(df['Total Revenue'] > 3000000) &
                 (df['Payment Method'] == 'Credit Card')]
print(f"\n🔍 High Value Credit Card Transactions:")
print(f"   Count: {len(high_credit)}")
print(high_credit[['Total Revenue', 'Payment Method', 'Region']].head())

# More than 10 units sold AND Unit Price > 100,000
bulk_expensive = df[(df['Units Sold'] > 10) & (df['Unit Price'] > 100000)]
print(f"\n🔍 Bulk & Expensive Products (Units > 10 AND Price > 100k):")
print(f"   Count: {len(bulk_expensive)}")
print(bulk_expensive[['Product Name', 'Units Sold', 'Unit Price', 'Total Revenue']].head())
```

```python
## Product Category Book & Region north america
df[(df['Product Category'] == 'Books') & (df['Region'] == 'North America')]
```

```python
# --- Exercise 2.3: Multiple Filters with OR (|) ---
print("\n--- Exercise 2.3: Multiple Filters with OR (|) ---")

# Jakarta OR Bandung
jakarta_bandung = df[(df['Region'] == 'Jakarta') | (df['Region'] == 'Bandung')]
print(f"\n🔍 Jakarta OR Bandung:")
print(f"   Count: {len(jakarta_bandung)}")
print(f"   Jakarta: {len(df[df['Region'] == 'Jakarta'])}")
print(f"   Bandung: {len(df[df['Region'] == 'Bandung'])}")

# Credit Card OR E-wallet
digital_payment = df[(df['Payment Method'] == 'Credit Card') |
                     (df['Payment Method'] == 'E-wallet')]
print(f"\n🔍 Credit Card OR E-wallet:")
print(f"   Count: {len(digital_payment)}")
```

```python
# --- Exercise 2.4: Complex Boolean Filtering ---
print("\n--- Exercise 2.4: Complex Boolean Filtering ---")

# (Electronics OR Clothing) AND (Jakarta OR Surabaya) AND Revenue > 2M
complex_filter = (
    ((df['Product Category'] == 'Electronics') | (df['Product Category'] == 'Clothing')) &
    ((df['Region'] == 'Jakarta') | (df['Region'] == 'Surabaya')) &
    (df['Total Revenue'] > 2000000)
)

filtered_data = df[complex_filter]
print(f"\n🔍 Complex Filter Results:")
print(f"   (Electronics OR Clothing) AND")
print(f"   (Jakarta OR Surabaya) AND")
print(f"   (Revenue > USD 2,000,000)")
print(f"\n   Count: {len(filtered_data)}")
print(filtered_data[['Product Category', 'Region', 'Total Revenue']].head(10))
```

```python
# --- Exercise 2.5: Using .isin() ---
print("\n--- Exercise 2.5: Filtering with .isin() ---")

# Multiple regions
selected_regions = df[df['Region'].isin(['Jakarta', 'Bandung', 'Surabaya'])]
print(f"\n🔍 Selected Regions (Jakarta, Bandung, Surabaya):")
print(f"   Count: {len(selected_regions)}")
print(selected_regions['Region'].value_counts())

# Multiple categories
selected_categories = df[df['Product Category'].isin(['Electronics', 'Clothing'])]
print(f"\n🔍 Selected Categories (Electronics, Clothing):")
print(f"   Count: {len(selected_categories)}")
print(selected_categories['Product Category'].value_counts())
```

```python
# --- Exercise 2.6: Using .loc[] ---
print("\n--- Exercise 2.6: Filtering with .loc[] ---")

# Select specific columns with filter
jakarta_info = df.loc[df['Region'] == 'Jakarta',
                      ['Date', 'Product Category', 'Total Revenue', 'Payment Method']]
print(f"\n🔍 Jakarta Sales Information:")
print(jakarta_info.head(10))
```

## Grouping & Aggregation

```python
# --- Exercise 3.1: Simple Grouping ---
print("\n--- Exercise 3.1: Simple Grouping with Single Aggregation ---")

# Total revenue by Product Category
revenue_by_category = df.groupby('Product Category')['Total Revenue'].sum()
print(f"\n Total Revenue by Product Category:")
print(revenue_by_category.sort_values(ascending=False))

# Average revenue by Region
avg_revenue_by_region = df.groupby('Region')['Total Revenue'].mean()
print(f"\n Average Revenue by Region:")
print(avg_revenue_by_region.sort_values(ascending=False).round(2))

# Total units sold by Payment Method
units_by_payment = df.groupby('Payment Method')['Units Sold'].sum()
print(f"\n Total Units Sold by Payment Method:")
print(units_by_payment.sort_values(ascending=False))
```

```python
# --- Exercise 3.2: Grouping with reset_index() ---
print("\n--- Exercise 3.2: Grouping with reset_index() ---")

# Convert to DataFrame
revenue_df = df.groupby('Product Category')['Total Revenue'].sum().reset_index()
revenue_df.columns = ['Product Category', 'Total Revenue']
revenue_df = revenue_df.sort_values('Total Revenue', ascending=False)

print(f"\n Revenue by Category (as DataFrame):")
print(revenue_df)
```

```python
# --- Exercise 3.3: Multiple Aggregations ---
print("\n--- Exercise 3.3: Multiple Aggregations on Single Column ---")

# Multiple stats for Total Revenue by Category
category_stats = df.groupby('Product Category')['Total Revenue'].agg([
    'sum', 'mean', 'min', 'max', 'count'
])
category_stats.columns = ['Total', 'Average', 'Min', 'Max', 'Count']
print(f"\n Revenue Statistics by Product Category:")
print(category_stats.round(2))

# Multiple stats by Region
region_stats = df.groupby('Region')['Total Revenue'].agg([
    ('Total_Revenue', 'sum'),
    ('Avg_Revenue', 'mean'),
    ('Transaction_Count', 'count')
])
print(f"\n Sales Statistics by Region:")
print(region_stats.round(2))
```

```python
# --- Exercise 3.4: Multiple Aggregations on Multiple Columns ---
print("\n--- Exercise 3.4: Aggregating Multiple Columns ---")

# Comprehensive analysis by Category
category_analysis = df.groupby('Product Category').agg({
    'Total Revenue': ['sum', 'mean'],
    'Units Sold': ['sum', 'mean'],
    'Transaction ID': 'count',
    'Unit Price': 'mean'
})

print(f"\n Comprehensive Category Analysis:")
print(category_analysis.round(2))
```

```python
# --- Exercise 3.5: Custom Named Aggregations ---
print("\n--- Exercise 3.5: Custom Named Aggregations ---")

sales_summary = (
    df.groupby('Product Category')
    .agg(
        Total_Revenue=('Total Revenue', 'sum'),
        Avg_Revenue=('Total Revenue', 'mean'),
        Total_Units=('Units Sold', 'sum'),
        Avg_Units=('Units Sold', 'mean'),
        Transaction_Count=('Transaction ID', 'count'),
        Avg_Price=('Unit Price', 'mean')
    )
    .round(2)
    .sort_values('Total_Revenue', ascending=False)
)

print(f"\n Sales Summary by Product Category:")
print(sales_summary)
```

```python
# --- Exercise 3.6: Grouping by Multiple Columns ---
print("\n--- Exercise 3.6: Grouping by Multiple Columns ---")

# Sales by Category and Region
category_region = (
    df.groupby(['Product Category', 'Region'])
    .agg(
        Total_Revenue=('Total Revenue', 'sum'),
        Transaction_Count=('Transaction ID', 'count'),
        Avg_Revenue=('Total Revenue', 'mean')
    )
    .reset_index()
    .sort_values('Total_Revenue', ascending=False)
)

print(f"\n Sales by Category and Region (Top 15):")
print(category_region.head(15))

# Sales by Region and Payment Method
region_payment = (
    df.groupby(['Region', 'Payment Method'])
    ['Total Revenue']
    .sum()
    .reset_index()
    .sort_values('Total Revenue', ascending=False)
)

print(f"\n Sales by Region and Payment Method:")
print(region_payment.head(10))
```

```python
# --- Exercise 3.7: Advanced Grouping ---
print("\n--- Exercise 3.7: Advanced Grouping Analysis ---")

# Monthly sales analysis
monthly_sales = (
    df.groupby('Month_Name')
    .agg(
        Total_Revenue=('Total Revenue', 'sum'),
        Total_Transactions=('Transaction ID', 'count'),
        Total_Units=('Units Sold', 'sum'),
        Avg_Transaction_Value=('Total Revenue', 'mean')
    )
    .round(2)
)

print(f"\n Monthly Sales Analysis:")
print(monthly_sales)

# Best performing products by category
best_products = (
    df.groupby(['Product Category', 'Product Name'])
    ['Total Revenue']
    .sum()
    .reset_index()
    .sort_values('Total Revenue', ascending=False)
)

print(f"\n Top 10 Products by Revenue:")
print(best_products.head(10))
```

```python
# --- Exercise 4.3: Sorting by Multiple Columns ---
print("\n--- Exercise 4.3: Sorting by Multiple Columns ---")

# Sort by Category (asc) then Revenue (desc)
multi_sort = df.sort_values(['Product Category', 'Total Revenue'],
                            ascending=[True, False])
print(f"\n Sorted by Category and Revenue:")
print(multi_sort[['Product Category', 'Product Name', 'Total Revenue']].head(15))

# Sort by Region (asc), Payment Method (asc), Revenue (desc)
complex_sort = df.sort_values(['Region', 'Payment Method', 'Total Revenue'],
                              ascending=[True, True, False])
print(f"\n Sorted by Region, Payment, and Revenue:")
print(complex_sort[['Region', 'Payment Method', 'Total Revenue']].head(15))
```

## Pivot Tables

```python
# --- Exercise 5.1: Simple Pivot Table ---
print("\n--- Exercise 5.1: Simple Pivot Table ---")

# Average revenue by Category and Region
pivot_simple = df.pivot_table(
    values='Total Revenue',
    index='Product Category',
    columns='Region',
    aggfunc='mean'
)

print(f"\n📊 Average Revenue by Category and Region:")
print(pivot_simple.round(2))
```

```python
# --- Exercise 5.2: Pivot with Sum ---
print("\n--- Exercise 5.2: Pivot Table with Sum ---")

# Total revenue by Category and Payment Method
pivot_sum = df.pivot_table(
    values='Total Revenue',
    index='Product Category',
    columns='Payment Method',
    aggfunc='sum',
    fill_value=0
)

print(f"\n📊 Total Revenue by Category and Payment Method:")
print(pivot_sum.round(2))
```

```python
# --- Exercise 5.3: Pivot with Multiple Values ---
print("\n--- Exercise 5.3: Pivot with Multiple Metrics ---")

# Revenue and Units by Category and Region
pivot_multi = df.pivot_table(
    values=['Total Revenue', 'Units Sold'],
    index='Product Category',
    columns='Region',
    aggfunc={'Total Revenue': 'sum', 'Units Sold': 'sum'},
    fill_value=0
)

print(f"\n📊 Revenue and Units by Category and Region:")
print(pivot_multi)
```

```python
# --- Exercise 5.4: Pivot with Margins (Totals) ---
print("\n--- Exercise 5.4: Pivot with Row and Column Totals ---")

# Revenue with totals
pivot_margins = df.pivot_table(
    values='Total Revenue',
    index='Product Category',
    columns='Region',
    aggfunc='sum',
    fill_value=0,
    margins=True,
    margins_name='TOTAL'
)

print(f"\n📊 Revenue by Category and Region (with Totals):")
print(pivot_margins.round(2))
```

```python
# --- Exercise 5.5: Complex Pivot ---
print("\n--- Exercise 5.5: Complex Pivot Table ---")

# Multi-level pivot
pivot_complex = df.pivot_table(
    values='Total Revenue',
    index=['Product Category', 'Payment Method'],
    columns='Region',
    aggfunc=['sum', 'mean', 'count'],
    fill_value=0
)

print(f"\n📊 Complex Pivot (Category & Payment vs Region):")
print(pivot_complex.head(20))
```

```python
# --- Exercise 5.6: Monthly Pivot ---
print("\n--- Exercise 5.6: Time-based Pivot Table ---")

# Monthly revenue by Category
pivot_monthly = df.pivot_table(
    values='Total Revenue',
    index='Month_Name',
    columns='Product Category',
    aggfunc='sum',
    fill_value=0
)

# Reorder months
month_order = ['January', 'February', 'March', 'April', 'May', 'June',
               'July', 'August', 'September', 'October', 'November', 'December']
pivot_monthly = pivot_monthly.reindex([m for m in month_order if m in pivot_monthly.index])

print(f"\n📊 Monthly Revenue by Product Category:")
print(pivot_monthly.round(2))

print("\n" + "=" * 60)
print("💡 PIVOT TABLE BENEFITS:")
print("   ✓ Quickly summarize data from multiple angles")
print("   ✓ Easy comparison across categories")
print("   ✓ Add totals with margins=True")
print("=" * 60)
```

## Cross Tabulation

```python
# --- Exercise 6.1: Simple Crosstab ---
print("\n--- Exercise 6.1: Simple Cross Tabulation ---")

# Count of transactions by Category and Region
crosstab_simple = pd.crosstab(df['Product Category'], df['Region'])
print(f"\n📊 Transaction Count by Category and Region:")
print(crosstab_simple)
```

```python
# --- Exercise 6.2: Crosstab with Margins ---
print("\n--- Exercise 6.2: Crosstab with Totals ---")

# With row and column totals
crosstab_margins = pd.crosstab(
    df['Product Category'],
    df['Region'],
    margins=True,
    margins_name='TOTAL'
)

print(f"\n📊 Transaction Count (with Totals):")
print(crosstab_margins)
```

```python
# --- Exercise 6.3: Crosstab with Percentages ---
print("\n--- Exercise 6.3: Crosstab with Percentages ---")

# Row percentages (% of each category in each region)
crosstab_row_pct = pd.crosstab(
    df['Product Category'],
    df['Region'],
    normalize='index'
) * 100

print(f"\n📊 Category Distribution by Region (Row %):")
print(crosstab_row_pct.round(2))

# Column percentages (% of each region's transactions by category)
crosstab_col_pct = pd.crosstab(
    df['Product Category'],
    df['Region'],
    normalize='columns'
) * 100

print(f"\n📊 Region Distribution by Category (Column %):")
print(crosstab_col_pct.round(2))

# Overall percentages
crosstab_all_pct = pd.crosstab(
    df['Product Category'],
    df['Region'],
    normalize='all'
) * 100

print(f"\n📊 Overall Distribution (%):")
print(crosstab_all_pct.round(2))
```

```python
# --- Exercise 6.4: Crosstab with Values ---
print("\n--- Exercise 6.4: Crosstab with Aggregated Values ---")

# Total revenue by Category and Payment Method
crosstab_revenue = pd.crosstab(
    df['Product Category'],
    df['Payment Method'],
    values=df['Total Revenue'],
    aggfunc='sum'
)

print(f"\n📊 Total Revenue by Category and Payment Method:")
print(crosstab_revenue.fillna(0).round(2))

# Average revenue by Region and Payment Method
crosstab_avg = pd.crosstab(
    df['Region'],
    df['Payment Method'],
    values=df['Total Revenue'],
    aggfunc='mean'
)

print(f"\n📊 Average Revenue by Region and Payment Method:")
print(crosstab_avg.fillna(0).round(2))
```

```python
# --- Exercise 6.5: Multi-level Crosstab ---
print("\n--- Exercise 6.5: Multi-level Cross Tabulation ---")

# Crosstab with multiple row variables
crosstab_multi = pd.crosstab(
    [df['Product Category'], df['Payment Method']],
    df['Region'],
    margins=True
)

print(f"\n📊 Multi-level Crosstab (Category & Payment vs Region):")
print(crosstab_multi.head(20))
```

```python
# --- Exercise 6.6: Payment Method Analysis ---
print("\n--- Exercise 6.6: Payment Method Preference by Category ---")

payment_category = pd.crosstab(
    df['Payment Method'],
    df['Product Category'],
    normalize='columns'
) * 100

print(f"\n📊 Payment Method Preference by Category (%):")
print(payment_category.round(2))

print("\n" + "=" * 60)
print("💡 CROSSTAB vs PIVOT:")
print("   CROSSTAB: Best for frequency counts and categorical analysis")
print("   PIVOT: Best for aggregating numerical values")
print("=" * 60)
```

## Working with Data Frame




```python
# Date column already converted, let's explore it
print("\n--- Exercise 8.1: Date Components Extraction ---")

# Extract various date components
df['Year'] = df['Date'].dt.year
df['Quarter'] = df['Date'].dt.quarter
df['Week'] = df['Date'].dt.isocalendar().week
df['Day'] = df['Date'].dt.day
df['Day_of_Week'] = df['Date'].dt.dayofweek  # Monday=0, Sunday=6
df['Is_Weekend'] = df['Day_of_Week'].isin([5, 6]).astype(int)
df['Day_of_Year'] = df['Date'].dt.dayofyear

print(f"\n📅 Date Components (first 10 rows):")
print(df[['Date', 'Year', 'Quarter', 'Month_Name', 'Day', 'Day_Name', 'Week', 'Is_Weekend']].head(10))
```

```python
# --- Exercise 8.2: Weekday vs Weekend Analysis ---
print("\n--- Exercise 8.2: Weekday vs Weekend Sales Analysis ---")

weekday_weekend = df.groupby('Is_Weekend').agg({
    'Total Revenue': ['sum', 'mean', 'count'],
    'Units Sold': 'sum'
})

weekday_weekend.index = ['Weekday', 'Weekend']
print(f"\n📅 Weekday vs Weekend Comparison:")
print(weekday_weekend.round(2))

# Sales by day of week
day_analysis = (
    df.groupby('Day_Name')
    .agg(
        Total_Revenue=('Total Revenue', 'sum'),
        Avg_Revenue=('Total Revenue', 'mean'),
        Transaction_Count=('Transaction ID', 'count')
    )
    .round(2)
)

# Reorder by day of week
day_order = ['Monday', 'Tuesday', 'Wednesday', 'Thursday', 'Friday', 'Saturday', 'Sunday']
day_analysis = day_analysis.reindex([d for d in day_order if d in day_analysis.index])

print(f"\n📅 Sales by Day of Week:")
print(day_analysis)
```

```python
# --- Exercise 8.3: Monthly Trends ---
print("\n--- Exercise 8.3: Monthly Sales Trends ---")

monthly_trend = (
    df.groupby('Month_Name')
    .agg(
        Total_Revenue=('Total Revenue', 'sum'),
        Avg_Revenue=('Total Revenue', 'mean'),
        Total_Transactions=('Transaction ID', 'count'),
        Total_Units=('Units Sold', 'sum')
    )
    .round(2)
)

print(f"\n📅 Monthly Sales Trend:")
print(monthly_trend)
```

```python
# --- Exercise 9.1: Understanding Data Scale ---
print("\n--- Exercise 9.1: Understanding Different Scales ---")

# Select numerical features
numerical_features = ['Units Sold', 'Unit Price', 'Total Revenue']

print(f"\n📊 Original Data Statistics:")
print(df[numerical_features].describe().round(2))

print(f"\n💡 Notice the different scales:")
print(f"   Units Sold: {df['Units Sold'].min()} - {df['Units Sold'].max()}")
print(f"   Unit Price: {df['Unit Price'].min():.2f} - {df['Unit Price'].max():.2f}")
print(f"   Total Revenue: {df['Total Revenue'].min():.2f} - {df['Total Revenue'].max():.2f}")
```

```python
# --- Exercise 9.2: StandardScaler (Z-Score Normalization) ---
print("\n--- Exercise 9.2: StandardScaler (Mean=0, Std=1) ---")

# Create copy for StandardScaler
df_standard = df.copy()

# Initialize and fit StandardScaler
scaler_standard = StandardScaler()
df_standard[numerical_features] = scaler_standard.fit_transform(df[numerical_features])

print(f"\n📊 After StandardScaler:")
print(df_standard[numerical_features].describe().round(2))

print(f"\n💡 StandardScaler Characteristics:")
print(f"   - Mean ≈ 0")
print(f"   - Standard Deviation = 1")
print(f"   - Preserves outliers")
print(f"   - Good for normally distributed data")

# Compare original vs scaled
comparison_standard = pd.DataFrame({
    'Original_Revenue': df['Total Revenue'].head(10),
    'Scaled_Revenue': df_standard['Total Revenue'].head(10),
    'Original_Units': df['Units Sold'].head(10),
    'Scaled_Units': df_standard['Units Sold'].head(10)
})

print(f"\n📊 Comparison (first 10 transactions):")
print(comparison_standard.round(3))
```

```python
# --- Exercise 9.3: MinMaxScaler (0-1 Normalization) ---
print("\n--- Exercise 9.3: MinMaxScaler (Range: 0-1) ---")

# Create copy for MinMaxScaler
df_minmax = df.copy()

# Initialize and fit MinMaxScaler
scaler_minmax = MinMaxScaler()
df_minmax[numerical_features] = scaler_minmax.fit_transform(df[numerical_features])

print(f"\n📊 After MinMaxScaler:")
print(df_minmax[numerical_features].describe().round(2))

print(f"\n💡 MinMaxScaler Characteristics:")
print(f"   - Min = 0")
print(f"   - Max = 1")
print(f"   - All values in range [0, 1]")
print(f"   - Sensitive to outliers")

# Compare original vs scaled
comparison_minmax = pd.DataFrame({
    'Original_Revenue': df['Total Revenue'].head(10),
    'MinMax_Revenue': df_minmax['Total Revenue'].head(10),
    'Original_Units': df['Units Sold'].head(10),
    'MinMax_Units': df_minmax['Units Sold'].head(10)
})

print(f"\n📊 Comparison (first 10 transactions):")
print(comparison_minmax.round(3))
```

```python
# --- Exercise 9.4: Side-by-Side Comparison ---
print("\n--- Exercise 9.4: StandardScaler vs MinMaxScaler ---")

comparison_both = pd.DataFrame({
    'Original_Revenue': df['Total Revenue'].head(10),
    'StandardScaler': df_standard['Total Revenue'].head(10),
    'MinMaxScaler': df_minmax['Total Revenue'].head(10)
})

print(f"\n📊 Three-Way Comparison (Total Revenue):")
print(comparison_both.round(3))
```
