## Task 1: Matplotlib Line Plot

**Objective:** Create a line plot to visualize time-series data (monthly sales trend).

**When to use:** Line plots are ideal for showing trends over time, continuous data, or relationships between two continuous variables.

```python
months = ['Jan', 'Feb', 'Mar', 'Apr', 'May', 'Jun', 'Jul', 'Aug', 'Sep', 'Oct', 'Nov', 'Dec']
sales = [1200, 1350, 1100, 1400, 1600, 1800, 1750, 1900, 2100, 2300, 2500, 3000]

plt.figure(figsize=(12, 6))
plt.plot(months, sales, color='teal', marker='o', linewidth=2, markersize=8)
plt.title("Monthly Sales Trend 2025", fontsize=16, fontweight='bold')
plt.xlabel("Month", fontsize=12)
plt.ylabel("Sales ($)", fontsize=12)
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
```

## Task 3: Distribution Plot

**Objective:** Visualize the distribution of a continuous variable (ticket fares).

**When to use:** Histograms with KDE (Kernel Density Estimation) curves help understand the shape, spread, and central tendency of data distributions.

```python
plt.figure(figsize=(10, 6))
sns.histplot(data=df, x='fare', kde=True, bins=30, color='coral')
plt.title("Distribution of Ticket Fares", fontsize=14)
plt.xlabel("Fare ($)")
plt.show()
```

## Task 4: Box Plot

**Objective:** Compare distributions across categories using box plots.

**When to use:** Box plots are perfect for comparing distributions, identifying outliers, and understanding the spread and quartiles of data across different groups.

```python
plt.figure(figsize=(8, 6))
sns.boxplot(data=df, x='pclass', y='age', palette='pastel')
plt.title("Age Distribution by Passenger Class", fontsize=14)
plt.show()
```

```python
plt.figure(figsize=(8, 6))
sns.boxplot(data=df, x='fare')
plt.show()
```

## Task 5: Plotly Interactive Scatter Plot

**Objective:** Create an interactive scatter plot with hover information.

**When to use:** Scatter plots reveal relationships between two continuous variables. Plotly adds interactivity for detailed data exploration.

```python
fig = px.scatter(df, x='age', y='fare', color='survived',
                 hover_data=['pclass', 'sex', 'embarked'],
                 title="Age vs Fare (Colored by Survival)",
                 labels={'survived': 'Survived'})
fig.show()
```

## Task 6: Interactive Histogram with Marginal Plots

**Objective:** Create a histogram with additional marginal visualizations.

**When to use:** Marginal plots provide a compact way to show distributions while also revealing patterns by categories.

```python
fig = px.histogram(
    df,
    x='age',
    color='sex',
    title="Age Distribution by Gender")
fig.show()
```

## Task 7: Interactive Box Plot with All Points

**Objective:** Create an interactive box plot that displays all individual data points.

**When to use:** Combining box plots with individual points helps identify the actual data distribution and outliers while maintaining statistical summaries.

```python
fig = px.box(df, x='pclass', y='fare', color='pclass', points='all',
             title="Fare Distribution by Passenger Class")
fig.show()
```

## Bonus: Correlation Heatmap

**Objective:** Visualize correlations between all numerical variables in the dataset.

**When to use:** Heatmaps are essential for understanding relationships between multiple variables simultaneously, especially in exploratory data analysis and feature selection for machine learning.

```python
corr = df.select_dtypes(include='number').corr()
fig = px.imshow(corr, text_auto='.2f', color_continuous_scale='RdBu_r',
                title="Correlation Heatmap (Titanic Dataset)")
fig.show()
```

### Explanation:

**Code Breakdown:**
- `df.select_dtypes(include='number')`: Selects only numerical columns
- `.corr()`: Calculates correlation matrix (Pearson correlation by default)
- `px.imshow()`: Creates a heatmap visualization
  - `text_auto='.2f'`: Displays correlation values with 2 decimal places
  - `color_continuous_scale='RdBu_r'`: Uses red-blue diverging color scale
    - Blue = positive correlation
    - Red = negative correlation
    - White = no correlation (0)

**Understanding Correlations:**
- **Range**: Correlation values range from -1 to +1
- **+1**: Perfect positive correlation (both increase together)
- **-1**: Perfect negative correlation (one increases, other decreases)
- **0**: No linear relationship
- **Diagonal**: Always 1 (each variable perfectly correlates with itself)

**Key Insights from Titanic Data:**
- `survived` and `pclass`: Negative correlation (lower class number = higher survival)
- `pclass` and `fare`: Negative correlation (lower class = lower fare)
- `survived` and `fare`: Positive correlation (higher fare = better survival)
- `age` and `pclass`: Weak positive correlation

**Machine Learning Application:**
- Identify which features strongly correlate with target variable (`survived`)
- Detect multicollinearity (highly correlated features that might cause issues)
- Guide feature selection and engineering decisions

**Best Practices:**
- Use diverging color scales for correlations (makes positive/negative clear)
- Include correlation values for precise interpretation
- Remember: Correlation ≠ Causation!

## Summary: Choosing the Right Visualization

| Visualization Type | Best Used For | Library | Key Advantage |
|-------------------|---------------|---------|---------------|
| **Line Plot** | Time series, trends over time | Matplotlib | Simple, clear trend visualization |
| **Bar Plot** | Comparing categories, showing differences | Seaborn | Easy comparison, statistical summaries |
| **Histogram** | Distribution of continuous data | Seaborn | Understand data spread and shape |
| **Box Plot** | Comparing distributions, finding outliers | Seaborn/Plotly | Compact statistical summary |
| **Scatter Plot** | Relationships between two variables | Plotly | Reveal correlations and patterns |
| **Heatmap** | Correlations, matrix data | Plotly/Seaborn | Show many relationships at once |

## Key Takeaways for Bootcamp Students

1. **Choose the right tool:**
   - **Matplotlib**: Fine-grained control, publication-quality static plots
   - **Seaborn**: Statistical visualizations with beautiful defaults
   - **Plotly**: Interactive, web-ready visualizations for exploration

2. **Always ask yourself:**
   - What story am I trying to tell?
   - Who is my audience?
   - What insights should they gain?
   - Is this data categorical, continuous, or both?

3. **Best practices:**
   - Always label axes with units
   - Use titles that describe the insight, not just the data
   - Choose color palettes carefully (consider colorblind accessibility)
   - Add context through annotations when needed
   - Keep it simple - avoid chart junk

4. **Common mistakes to avoid:**
   - Using 3D plots when 2D would suffice
   - Too many colors or categories in one plot
   - Missing axis labels or units
   - Misleading scales (not starting y-axis at 0 for bar charts)
   - Choosing the wrong chart type for your data
